## **Lista 1:** Introdução ao Pandas

```python
import pandas as pd
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/joaohsantanaoc/COLTEC/refs/heads/Python/Domotica_Robotica/cursos-prouni.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.drop(columns=['curso_id', 'cidade_filtro', 'campus_id'], inplace=True)
```

```python
df.head()
```

#### **Questão 1:** 10 cursos com maior mensalidade

```python
top = df.nlargest(10, 'mensalidade')
top[['grau', 'turno', 'mensalidade', 'curso_busca', 'cidade_busca', 'uf_busca', 'universidade_nome', 'campus_nome', 'nome']]
```

#### **Questão 2:** Curso com maior diferença entre a nota de ampla para cotas

```python
diferenca = df['nota_integral_ampla'] - df['nota_integral_cotas']
index = diferenca.idxmax()
df.loc[index, ['grau', 'turno', 'mensalidade', 'curso_busca', 'cidade_busca', 'uf_busca', 'universidade_nome', 'campus_nome', 'nome', 'nota_integral_ampla', 'nota_integral_cotas']]
```

#### **Questão 3:** Média das mensalidades para cada cidade e top 10 cidades com maior média

```python
media_mensalidade = df.groupby('cidade_busca')['mensalidade'].mean()
media_mensalidade.sort_values(ascending=False).head(10)
```

#### **Questão 4:** Curso mais caro para o governo:

```python
curso_mais_caro_ampla = df['bolsa_integral_ampla'] * df['mensalidade'] + df['bolsa_parcial_ampla'] * df['mensalidade'] / 2
index_ampla = curso_mais_caro_ampla.idxmax()
df.loc[index_ampla, ['grau', 'turno', 'mensalidade', 'curso_busca', 'cidade_busca', 'uf_busca', 'universidade_nome', 'campus_nome', 'nome', 'bolsa_integral_ampla', 'bolsa_parcial_ampla']]
```

#### **Questão 5:** Histograma do valor da mensalidade para medicina

```python
import matplotlib.pyplot as plt
```

```python
df.info()
```

```python
cursos_medicina = df[df['nome'].str.contains('Medicina')]
mensalidades_medicina = cursos_medicina['mensalidade']

plt.hist(mensalidades_medicina, bins=30, range=(0, 10000))

plt.title('Histograma da mensalidade para medicina')
plt.xlabel('Mensalidade')
plt.ylabel('Número de cursos')
plt.show()
```
