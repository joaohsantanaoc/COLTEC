```python
# Importando a Biblioteca Pandas
import pandas as pd
```

```python
# Acessando nosso DataFrame via GitHub (também é possível pela localização do arquivo):
df = pd.read_csv("https://raw.githubusercontent.com/joaohsantanaoc/COLTEC/refs/heads/Python/Domotica_Robotica/cursos-prouni.csv")
```

```python
# Visualizando os tipos de dados presentes:
df.info()
```

```python
# Visualizando as colunas:
df.columns
```

```python
# Removendo as colunas desnecessárias:
df.drop(columns=['curso_id', 'cidade_filtro', 'campus_id'], inplace=True)
```

```python
# Visualizando nosso DataFrame:
df.head()
```

##### **Questão 1:** Top 10 cursos com maior mensalidade

```python
# Usamos a função nlarget para ver os 10 cursos com maior mensalidade:
top = df.nlargest(10, ['mensalidade'])
top[['grau', 'turno', 'mensalidade', 'curso_busca', 'cidade_busca', 'uf_busca', 'universidade_nome', 'campus_nome']]
```

##### **Questão 2:** Curso com maior diferença para a nota de ampla para cotas

```python
# Obtemos a diferença entre as notas de corte para todos os cursos e depois achamos o indíce de maior diferença:
diferenca = df['nota_integral_ampla'] - df['nota_integral_cotas']
index = diferenca.idxmax()
df.loc[index, ['grau', 'turno', 'mensalidade', 'curso_busca', 'cidade_busca', 'uf_busca', 'universidade_nome', 'campus_nome', 'nota_integral_ampla', 'nota_integral_cotas']]
```

##### **Questão 3:** Média de mensalidade para cidades e top 10 cidade com maior mensalidade

```python
# Agregamos os dados da cidade com mensalidade e depois ordenamos os valores em ordem crescente:
cidades_media_mensalidade = df.groupby('cidade_busca')['mensalidade'].mean()
cidades_media_mensalidade.sort_values(ascending=False, inplace=True)
cidades_media_mensalidade.head(10)
```

##### **Questão 4:** Curso mais caro para o governo

```python
# O curso mais caro para o governo é de ampla concorrência: (usei um if para verificar, mas por algum motivo não printou o curso)
curso_mais_caro = df['bolsa_integral_ampla'] * df['mensalidade'] + df['bolsa_parcial_ampla'] * df['mensalidade'] / 2
indice = curso_mais_caro.idxmax()
df.loc[indice, ['grau', 'turno', 'mensalidade', 'curso_busca', 'cidade_busca', 'uf_busca', 'universidade_nome', 'campus_nome', 'bolsa_integral_ampla', 'bolsa_parcial_ampla']]
```

##### **Questão 5:** Histograma Cursos Medicina

```python
# Importando a biblioteca matplotlib:
import matplotlib.pyplot as plt
```

```python
# Verificando os dados do df:
df.info()
```

```python
# Filtrando os cursos de medicina e suas mensalidades:
cursos_medicina = df[df['nome'].str.contains('Medicina')]
mensalidades_medicina = cursos_medicina['mensalidade']

plt.figure(1)
plt.hist(mensalidades_medicina, bins=30, range=(0, 10000))
plt.title('Histograma Cursos Medicina')
plt.xlabel('Mensalidade')
plt.ylabel('Quantidade de Cursos')
plt.show()
```
